### Why Convert Words to Vectors?

Understanding the meaning and relationships between words is crucial for many natural language processing tasks. However, computers understand numbers, not words. To bridge this gap and allow computers to work with the nuances of language, we use a process called **word vectorization** or **word embedding**.

By converting words into numerical representations called **vectors**, we transform the problem of finding semantic relationships (how words relate in meaning) into a problem of finding mathematical relationships between these vectors in a multi-dimensional space.

**In simpler terms:**

Imagine each word has a unique address in a vast city. Word vectorization gives each word a set of coordinates (the vector). Words that are related in meaning (like "king" and "queen," or "happy" and "joyful") will have coordinates that are close to each other in this city. This allows computers to easily calculate how similar words are or even understand analogies by looking at the distances and directions between their vector "addresses."

This process allows computers to understand and work with the meaning of words in a powerful way, enabling tasks like:

*   Finding synonyms and related words
*   Translating languages
*   Analyzing sentiment
*   Building better search engines
*   And much more!

**dot product of two vectors**

The dot product of two vectors is a common way to measure their similarity or how well they are aligned in the vector space.

* Large positive dot product: Indicates that the vectors are pointing in roughly the same direction and have significant magnitude. This **suggests the words they represent are highly similar or related**.

* Dot product close to zero: Suggests the vectors are nearly orthogonal (perpendicular) to each other. This **implies the words are largely unrelate**d or have very different meanings in the context of the model.

* Large negative dot product: Means the vectors are pointing in roughly opposite directions. This can **sometimes indicate words with opposite meanings**, though it's less common than positive correlations in typical word embedding spaces.

```python
import gensim.downloader

model = gensim.downloader.load('word2vec-google-news-300')
# model = gensim.downloader.load('glove-wiki-gigaword-50')
```

```python

# print("vector-->", model["man"])
# print("vector-->", model["king"])
print("distance between man and king", model.distance("man", "king"))
print("distance between man and women", model.distance("man", "women"))
print("distance between dog and cat", model.distance("dog", "cat"))
print("distance between cat and car", model.distance("cat", "car"))
print("words similar to cat", model.similar_by_word("cat"))
print("womenVect + (kingVect - manVect) should give =", model.most_similar_cosmul(positive=['king','women'], negative=['man']))
```

```python
import gensim.downloader
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt

# Load the model (using the same model as before)
# model = gensim.downloader.load('glove-wiki-gigaword-50')

# Hardcoded list of words to display
word_list_str = "uncle, aunt, man, woman, father, mother"
word_list = [word.strip() for word in word_list_str.split(',')]

# Extract vectors for the entered words
vectors = []
labels = []
for word in word_list:
    try:
        vectors.append(model[word])
        labels.append(word)
    except KeyError:
        print(f"Word '{word}' not found in the model vocabulary.")

if not vectors:
    print("No valid words entered to visualize.")
else:
    # Combine the vectors
    combined_vectors = np.vstack(vectors)

    # Apply PCA to reduce the dimensionality to 2
    pca = PCA(n_components=2)
    reduced_vectors = pca.fit_transform(combined_vectors)

    # Create a scatter plot
    plt.figure(figsize=(8, 6))
    plt.scatter(reduced_vectors[:, 0], reduced_vectors[:, 1])

    # Label the points
    for i, label in enumerate(labels):
        plt.annotate(label, (reduced_vectors[i, 0], reduced_vectors[i, 1]), textcoords="offset points", xytext=(0, 10), ha='center')

    plt.xlabel("PCA Component 1")
    plt.ylabel("PCA Component 2")
    plt.title("2D PCA Projection of Word Vectors")
    plt.grid(True)
    plt.show()
```
